# Matplotlib

## Importation de la bibliothèque

Souvent, on importe `matplotlib.pyplot` avec l'alias `plt`. Nous importons également NumPy pour générer des données d'exemple.

````python
import matplotlib.pyplot as plt
import numpy as np
from math import exp #on importe la fonction exponentielle
````

## Tracé Basique

La fonction `plt.plot()` est la méthode fondamentale pour créer des graphiques linéaires. Elle prend en arguments des séquences de coordonnées x et y.

````python
# Créer des données d'exemple
X_mpl = np.linspace(1, 10, 100)
Y_mpl = [exp(elt) for elt in X_mpl]

# Tracé basique
plt.plot(X_mpl, Y_mpl)
plt.show() # Affiche le graphique
````

## Personnalisation des graphiques

Ajoutez du contexte à vos graphiques avec des titres et des étiquettes d'axes. Contrôlez la plage visible avec les limites.



````python
# Tracer les données
plt.plot(X_mpl, Y_mpl)

# Ajouter titre et labels
plt.title('Fonction exponentielle')
plt.xlabel('Axe X')
plt.ylabel('Axe Y')

# Définir les limites des axes
plt.xlim(8, 10)
plt.ylim(5000, 15000)

plt.show()
````

Modifiez l'apparence de vos lignes et ajoutez des marqueurs.

- Couleurs (`color`): Utilisez des noms ('red', 'blue'), des codes hexadécimaux ('#FF8C00'), ou des raccourcis ('r', 'b', 'g').

- Styles de Ligne (`linestyle` ou `ls`): Styles courants : '-' (solide), '--' (tirets), ':' (pointillés), '-.' (tiret-point).

- Marqueurs (`marker`): Ajoutez des symboles aux points de données (ex : 'o', 'x', '+', 's', '^').

````python
# Préparation de nouvelles données
X_sin = np.linspace(0, 2 * np.pi, 50)
Y_sin = np.sin(X_sin)
Y_cos = np.cos(X_sin)

# Tracer avec des styles et marqueurs personnalisés
plt.figure(figsize=(10, 5)) # Définir la taille de la figure
plt.plot(X_sin, Y_sin, color='#FF8C00', ls=':', marker='o', label='Sinus (Pointillé Orange)') # Ligne pointillée orange avec marqueurs cercles
plt.plot(X_sin, Y_cos, color='green', ls='--', marker='x', label='Cosinus (Tirets Verts)') # Ligne tirets verts avec marqueurs x
plt.title('Sinus et Cosinus avec Styles')
plt.xlabel('Angle (radians)')
plt.ylabel('Valeur')
plt.legend() # Afficher les labels définis dans plot()
plt.grid(True) # Ajouter une grille
plt.show()
````

# Autres Types de Graphiques Courants

Bien que plt.plot soit polyvalent, pyplot propose des fonctions dédiées pour les types de graphiques courants :     

- Nuages de Points (`plt.scatter`): Idéal pour montrer la relation entre deux variables numériques.

- Diagrammes en Barres (`plt.bar`): Utile pour comparer des quantités entre différentes catégories.

- Histogrammes (`plt.hist`): Montre la distribution d'une seule variable numérique.

- Diagrammes en Boîte (Box Plots) (`plt.boxplot`): Résume la distribution des données à l'aide des quartiles, de la médiane et identifie les valeurs aberrantes (outliers).


# Figures Multiples (Subplots)



````python
# Créer une grille 2x2 de subplots
fig, axes = plt.subplots(nrows=2, ncols=2, figsize=(10, 8))
fig.suptitle('Exemples de Subplots Matplotlib') # Titre global pour la figure

# Tracer sur chaque subplot (objet Axes)
axes[0, 0].plot(X_mpl, Y_mpl, color='blue', ls=':')
axes[0, 0].set_title('Plot (0,0)')

X_scatter = np.cos(np.linspace(-3, 3, 50))**3
Y_scatter = np.sin(np.linspace(-3, 3, 50))**5

axes[0, 1].scatter(X_scatter, Y_scatter, color='red', marker='o')
axes[0, 1].set_title('Scatter (0,1)')

data_rd = np.random.randint(0, 10, 500)

axes[1, 0].hist(data_rd, bins=20, color='green')
axes[1, 0].set_title('Histogram (1,0)')

data_rd2 = np.random.normal(5, 3, 300)

axes[1, 1].boxplot([data_rd, data_rd2], patch_artist=True)
axes[1, 1].set_title('Boxplot (1,1)')
# Combine les positions et les étiquettes de manière sécurisée
axes[1, 1].set_xticks([1, 2], labels=['G1', 'G2'])

# Améliorer la disposition pour éviter le chevauchement des titres/labels
plt.tight_layout(rect=[0, 0.03, 1, 0.95]) # rect ajuste la disposition par rapport au suptitle

plt.show()
````

## Tracé d'un scatter plot

La fonction `scatterplot` permet de tracer des nuages de points, idéaux pour visualiser la relation entre deux variables numériques.

- Tracer un nuage de points représentant les ventes (`sales`) en fonction du salaire (`salary`) :



````python
plt.figure(figsize=(12, 8)) # Ajuste la taille de la figure
sns.scatterplot(x='salary', y='sales', data=df)
plt.title('Ventes en fonction du Salaire') # Ajout d'un titre
plt.xlabel('Salaire') # Label axe X
plt.ylabel('Ventes') # Label axe Y
plt.grid(True) # Ajout d'une grille pour faciliter la lecture
plt.show() # Afficher le graphique
````

## Le paramètre `hue`

On peut ajouter une troisième dimension (une variable catégorielle) en colorant les points selon ses valeurs grâce au paramètre `hue`

- Tracer les ventes en fonction du salaire, colorés selon la `division` :

````python
plt.figure(figsize=(12, 8))
sns.scatterplot(x='salary', y='sales', data=df, hue='division') # Coloration par division
plt.title('Ventes vs Salaire par Division')
plt.xlabel('Salaire')
plt.ylabel('Ventes')
plt.legend(title='Division') # Ajout d'un titre à la légende
plt.grid(True)
plt.show()
````

Même question mais avec `work experience` comme variable pour `hue`. On peut aussi changer la palette de couleurs. [Liste des palettes](https://matplotlib.org/stable/users/explain/colors/colormaps.html)

````python
plt.figure(figsize=(12, 8))
sns.scatterplot(x="salary", y="sales", data=df, hue='work experience', palette='viridis') # Palette 'viridis'
plt.title('Ventes vs Salaire (Couleur par Expérience)')
plt.xlabel('Salaire')
plt.ylabel('Ventes')
plt.legend(title='Expérience (années)')
plt.grid(True)
plt.show()
````

## Le paramètre `size`

Permet de faire varier la taille des points selon les valeurs d'une autre variable (numérique ou catégorielle).

- Tracer les ventes en fonction du salaire, avec la taille des points selon `work experience` :

````python
plt.figure(figsize=(12, 8))
sns.scatterplot(x='salary', y='sales', data=df, size='work experience', sizes=(20, 200)) # sizes définit la plage de taille
plt.title('Ventes vs Salaire (Taille par Expérience)')
plt.xlabel('Salaire')
plt.ylabel('Ventes')
plt.legend(title='Expérience (années)', bbox_to_anchor=(1.05, 1), loc='upper left') # Légende à l'extérieur
plt.grid(True)
plt.tight_layout() # Ajuste pour la légende
plt.show()
````

## Le paramètre `style`

Permet de faire varier la forme des marqueurs selon une variable catégorielle.

- Tracer les ventes en fonction du salaire, avec un style de marqueur différent pour chaque division :

````python
plt.figure(figsize=(12, 8))
# s=100 augmente la taille de base des points pour mieux voir le style
sns.scatterplot(x='salary', y='sales', data=df, style="division", s=100)
plt.title('Ventes vs Salaire (Style par Division)')
plt.xlabel('Salaire')
plt.ylabel('Ventes')
plt.legend(title='Division')
plt.grid(True)
plt.show()
````

## Synthèse Scatterplot

On peut combiner `hue`, `size`, et `style` pour visualiser plusieurs dimensions.

Tracer les ventes vs salaire :

`hue` = `level of education`

`size` = `work experience`



````python
plt.figure(figsize=(14, 9))
sns.scatterplot(x='salary', y='sales', data=df,
                hue='level of education',
                size='work experience',
                style='training level', # Ajout du style pour une dimension de plus
                sizes=(30, 300),        # Plage de taille
                palette='muted')        # Autre palette
plt.title('Ventes vs Salaire (Couleur: Éducation, Taille: Expérience, Style: Entraînement)')
plt.xlabel('Salaire')
plt.ylabel('Ventes')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True)
plt.tight_layout()
plt.show()
````

# Les graphiques de distributions

## ```displot()```, ```histplot()```  et ```kdeplot()``` <a class="anchor" id="section_3_3_2"></a>

- `histplot()` : Trace un histogramme. Utile pour voir la fréquence des valeurs dans des intervalles (bins).

- `kdeplot()` : Trace une estimation de densité par noyau (KDE). C'est une version lissée de l'histogramme montrant la forme de la distribution.

- `displot()` : Une fonction plus générale ("figure-level") qui peut tracer les deux (`kind='hist'` ou `kind='kde'`) et plus, et permet facilement de créer des facettes (plusieurs graphiques).

- Tracer un histogramme simple pour `salary` :

````python
plt.figure(figsize=(8, 5)) # histplot est "axes-level", on peut utiliser plt.figure
sns.histplot(x="salary", data=df, bins=20) # bins contrôle le nombre de barres
plt.title('Distribution des Salaires (Histogramme)')
plt.show()
````

- Utiliser `displot` pour ajouter la courbe KDE :

````python
# displot est "figure-level", la taille se contrôle différemment
sns.displot(x='salary', data=df, kde=True, height=5, aspect=1.5) # kde=True ajoute la courbe
plt.title('Distribution des Salaires (Hist + KDE via displot)')
plt.show()
````

- Ajouter un `rugplot` sous la distribution avec `displot` :

````python
sns.displot(x='salary', data=df, kde=True, rug=True, color='green', height=5, aspect=1.5)
# rug présente les observations via des marques au dessus de l'axe X.
plt.title('Distribution des Salaires (KDE + Rug via displot)')
plt.show()
````

- Tracer seulement la courbe KDE avec `kdeplot` et remplir la zone (`fill=True`) :

````python
sns.kdeplot(x='salary', data=df, color='red')
plt.title('Distribution des Salaires (kdeplot)')
plt.show()
````

## Les graphiques de variables catégorielles (estimation) <a class="anchor" id="section_3_4"></a>

Ces graphiques montrent une estimation statistique (comme la moyenne ou le comptage) pour chaque catégorie.

## Le `countplot()` <a class="anchor" id="section_3_4_1"></a>
Compte le nombre d'occurrences pour chaque catégorie d'une variable.

Compter les occurrences pour `division` :

````python
plt.figure(figsize=(10, 6))
sns.countplot(x='division', data=df, palette='viridis') # palette change les couleurs
plt.title('Nombre d\'employés par Division')
plt.ylabel('Nombre') # L'axe Y est le comptage
plt.xlabel('Division')
plt.show()
````

- Compter les occurrences pour `level of education` :

````python
plt.figure(figsize=(12, 6))
# Utiliser y='...' pour un countplot horizontal, plus lisible avec beaucoup de catégories
sns.countplot(y='level of education', data=df, order = df['level of education'].value_counts().index) # Ordonner par fréquence
plt.title("Nombre d'employés par Niveau d'Éducation")
plt.xlabel('Nombre')
plt.ylabel("Niveau d'Éducation")
plt.show()
````

- Ajouter une distinction (`hue`) basée sur `training level` :

````python
plt.figure(figsize=(12, 7))
sns.countplot(x='level of education', hue='training level', data=df, palette='Set2')
plt.title("Niveau d'Éducation vs Niveau d'Entraînement")
plt.xlabel("Niveau d'Éducation")
plt.ylabel('Nombre')
plt.xticks(rotation=45, ha='right') # Rotation des labels de l'axe X
plt.tight_layout() # Ajuste la mise en page
plt.show()
````

## Le `barplot()` <a class="anchor" id="section_3_4_2"></a>
Affiche une statistique (par défaut, la moyenne) d'une variable numérique pour chaque catégorie, avec un intervalle de confiance.

- Afficher le salaire moyen (`salary`) pour chaque `level of education` :

````python
plt.figure(figsize=(12, 6))
# 'estimator=np.mean' est le comportement par défaut, errorbar='sd' montre l'écart-type au lieu de l'IC
sns.barplot(x='level of education', y='salary', data=df, palette='viridis',
            hue='level of education', legend=False, errorbar='sd',
            order=df.groupby('level of education')['salary'].mean().sort_values(ascending=False).index) # Ordonner par salaire moyen
plt.title('Salaire Moyen (et Écart-Type) par Niveau d\'Éducation')
plt.xlabel("Niveau d'Éducation")
plt.ylabel('Salaire Moyen')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
````

## ```Le boxplot()```  <a class="anchor" id="section_3_5_1"></a>

Le boxplot est un graphique très utile pour résumer la distribution d'une variable numérique. Il montre rapidement :

- La médiane (le trait au milieu de la boîte) : la valeur qui sépare les données en deux moitiés.

- Les quartiles (les bords de la boîte) :

- Le 1er quartile (Q1) : 25% des données sont en dessous.

- Le 3ème quartile (Q3) : 75% des données sont en dessous.

- La boîte contient donc les 50% "du milieu" des données.

- Les moustaches (les traits qui sortent de la boîte) : elles montrent généralement l'étendue des données "normales".

- Les points isolés (parfois appelés outliers) : ce sont des valeurs qui sont très éloignées du reste des données.

### Le ```pairplot()``` <a class="anchor" id="section_3_6_2"></a>

Le pairplot est un outil pour avoir une vue d'ensemble rapide des relations entre toutes les variables numériques de votre DataFrame. Il crée une grille de graphiques :

Sur la diagonale : Il montre la distribution de chaque variable individuelle (souvent sous forme d'histogramme ou de courbe de densité).

Hors de la diagonale : Il montre les nuages de points (scatter plots) entre chaque paire de variables. Cela permet de voir s'il y a des corrélations ou des tendances entre elles.

 <span class="text-muted"> Tracer le pairplot du data frame

# Plotly

Plotly dont la documentation se trouve [ici](https://plotly.com/python), permet de réaliser des visualisations graphiques interactives.

* <span class="text-muted"> Importer la bibliothèque ```plotly``` sous le nom ```px``` </span>

Plotly a des données d'exemple. On va utiliser gapminder, qui contient des infos sur les pays (population, espérance de vie, PIB) au fil des années.

* <span class="text-muted"> Importation du data set ```gapminder```</span>

````python
df5 = px.data.gapminder()
````

* Tracé de la population par continent au fil des années

````python
px.bar(df5, x="continent", y="pop", color="continent",
      animation_frame="year", animation_group="country", range_y=[0,4000000000])
````

Explication simple :

- Ce code crée un diagramme en barres.

- Chaque barre représente un continent (x="continent").

- La hauteur de la barre montre la population (y="pop").

- Chaque continent a sa propre couleur (color="continent").

- animation_frame="year" ajoute un curseur de temps (ou un bouton "Play") qui montre l'évolution année par année.

Maintenant, on veut voir s'il y a un lien entre la richesse d'un pays (PIB par habitant) et l'espérance de vie, et comment ça change dans le temps. On utilise ```px.scatter``` pour faire un nuage de points animé (diagramme à bulles).

* Tracé de l'espérance de vie en fonction du PIB par pays au fil des années

````python
px.scatter(df5, x="gdpPercap", y="lifeExp", animation_frame="year", animation_group="country",
           size="pop", color="continent", hover_name="country",
           log_x=True, size_max=55, range_x=[100,100000], range_y=[25,90])
````
